# PR-5: Temporal Reasoning Engine\n\n**Ninai SDK Practical Example**\n\nThis notebook demonstrates temporal reasoning using the Ninai Python SDK.

## 1. Setup

````python
from ninai import NinaiClient\nimport pandas as pd\nimport numpy as np\nimport matplotlib.pyplot as plt\nfrom datetime import datetime, timedelta\n\nclient = NinaiClient(api_key='dev-key-pr5', base_url='http://localhost:8000/api/v1')\nprint(' Ninai SDK initialized')\nprint(' client.temporal methods available')
````

## 2. Tag Temporal Facts

````python
# Tag fact with validity interval\nfact = client.temporal.tag_fact_validity(\n    fact_id='user_skill_python',\n    valid_from=(datetime.utcnow() - timedelta(days=30)).isoformat(),\n   valid_to=(datetime.utcnow() - timedelta(days=20)).isoformat(),\n    confidence_at_time=0.85,\n    change_type='stable'\n)\nprint(f'Tagged fact: {fact["fact_id"]}')
````

## 3. Detect Event Sequences

````python
# Detect recurring patterns\nsequence = {\n    'entity_timeline': [\n        {'entity_id': 'problem_reported', 'timestamp': (datetime.utcnow() - timedelta(days=30)).isoformat()},\n        {'entity_id': 'problem_resolved', 'timestamp': (datetime.utcnow() - timedelta(days=28)).isoformat()}\n    ],\n    'min_occurrences': 2\n}\nsequences = client.temporal.detect_sequences([sequence])\nprint(f'Detected {len(sequences)} sequences')
````

## 4. Compute Trajectories

````python
# Generate engagement data\nmeasurements = [{'timestamp': (datetime.utcnow() - timedelta(days=30-i)).isoformat(), 'value': 0.5 + 0.01*i} for i in range(30)]\n\ntrajectory_req = {'entity_id': 'user_engagement', 'quantity': 'engagement', 'measurements': measurements}\ntrajectories = client.temporal.compute_trajectories([trajectory_req])\nprint(f'Trend: {trajectories[0]["trend_direction"]}')
````

## 5. Forecast Future

````python
# Forecast next 7 periods\nforecast = client.temporal.forecast_trajectory(trajectories[0]['id'], horizon_periods=7)\nprint(f'Forecast {len(forecast["predictions"])} periods')
````

## 6. Detect Inflection Points

````python
# Find changes\ninflections = client.temporal.get_inflection_points('user_motivation', sensitivity=2.0)\nprint(f'Found {len(inflections)} inflection points')
````

## 7. Optimal Action Timing

````python
# Determine when to act\ntiming = client.temporal.when_should_act(\n    goal_context={'goal_type': 'maintain_engagement'},\n    trajectory={'entity_id': 'user_engagement', 'current_value': 0.65},\n    action_lead_time_hours=24\n)\nprint(f'Recommendation: {timing["recommendation"]}')
````
